### Как получить номер дня недели по дате в таблице?

```python
def get_weekday(x):
    return x.weekday()
```

```python
vacancy.vacdate.apply(lambda x: get_weekday(x))
```

```python
a = vacancy.vactitle == "Дизайнер мобильных приложений"
```

```python
vacancy[a].vacdate.dt.weekday
```

```python
vacancy['weekday'] = vacancy['vacdate'].apply(lambda x: x.weekday())
```

```python
# найти профессии с самым большим количеством вакансий и построить по ним график (столбиковую диаграмму количества вакансий)
vacs = vacancy.groupby('vactitle')['vacid'].count().reset_index()
```

## Как сортировать таблицу по определенным столбцам?

```python
vacs = vacs.sort_values(by='vacid',ascending=False)
```

### Как выбрать топ строк по наибольшим значениям определенной колонки?

```python
vacs.nlargest(5, 'vacid')
```

### Как выбрать топ строк по наименьшим значениям определенной колонки?

```python
vacs.nsmallest(5, 'vacid')
```

### Как быстро построить график в pandas?

```python
vacs.nlargest(5, 'vacid').set_index('vactitle').plot(kind='barh');
```
